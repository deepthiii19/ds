```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
iris= load_iris()
df= pd.DataFrame(iris.data,columns=iris.feature_names)
df['species']=iris.target

plt.figure(figsize=(6,4))
sns.histplot(df['sepal length (cm)'],bins=20,kde=True)
plt.title("Histogram of Sepal Length")
plt.show()

plt.figure(figsize=(6,4))
sns.boxplot(x='species',y='sepal length (cm)',data=df)
plt.title("Boxplot of Sepal Length by Species")
plt.show()

speices_counts=df['species'].value_counts()
plt.figure(figsize=(6,6))
plt.pie(speices_counts,labels=iris.target_names,autopct='%1.1f%%')
plt.title("Pie chart")
plt.show()

plt.figure(figsize=(6,4))
sns.scatterplot(x='sepal length (cm)',y='sepal width (cm)',hue='species',data=df)
plt.title("Scatterplot of Sepal Length vs Sepal Width")
plt.show()

plt.figure(figsize=(6,4))
plt.plot(df['sepal length (cm)'])
plt.title("Line chart of sepal length")
plt.xlabel("Sample")
plt.ylabel("Sepal Length")
plt.show()

plt.figure(figsize=(6,4))
sns.barplot(x='species',y='sepal length (cm)',data=df)
plt.title("Bar Chaart: Mean Sepal Length by Species")
plt.xticks([0,1,2],iris.target_names)

plt.show()


plt.figure(figsize=(8,6))
sns.heatmap(df.iloc[:,:4].corr(),annot=True,cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()

plt.figure(figsize=(6,4))
plt.scatter(df['sepal length (cm)'],df['sepal width (cm)']*20, alpha=0.5,c=df['species'])
plt.title("Bubble Chart: Sepal Length vs Sepal Width")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")

sns.pairplot(df,vars=iris.feature_names,hue='species')
plt.suptitle("Pair Plot of Iris Features", y=1.02)
plt.show()

plt.figure(figsize=(6,4))
sns.violinplot(x='species',y='sepal length (cm)',data=df)
plt.title("Violin Plot of Sepal Length by Species")
plt.show()
```

```python
import seaborn as sns

tips_df = sns.load_dataset('tips')

# Display the first 5 rows of the DataFrame
display(tips_df.head())
```

### Exploratory Data Analysis for Tips Dataset

```python

plt.figure(figsize=(8, 5))
sns.histplot(tips_df['total_bill'], bins=15, kde=True)
plt.title('Distribution of Total Bill')
plt.xlabel('Total Bill ($)')
plt.ylabel('Frequency')
plt.show()
```

```python

plt.figure(figsize=(8, 5))
sns.boxplot(x='day', y='tip', data=tips_df, order=['Thur', 'Fri', 'Sat', 'Sun'])
plt.title('Tip Amount by Day of the Week')
plt.xlabel('Day of the Week')
plt.ylabel('Tip Amount ($)')
plt.show()
```

```python

plt.figure(figsize=(8, 5))
sns.scatterplot(x='total_bill', y='tip', hue='sex', data=tips_df)
plt.title('Total Bill vs Tip by Sex')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip Amount ($)')
plt.show()
```

```python

plt.figure(figsize=(8, 5))
sns.countplot(x='day', data=tips_df, order=['Thur', 'Fri', 'Sat', 'Sun'])
plt.title('Number of Diners by Day of the Week')
plt.xlabel('Day of the Week')
plt.ylabel('Number of Diners')
plt.show()
```
