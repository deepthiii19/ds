```python
import pandas as pd
from sklearn.datasets import load_iris
from statistics import mode
iris= load_iris()
df=pd.DataFrame(iris.data, columns=iris.feature_names)
print("First 5 rows of dataset:")
print(df.head())
mean_values=df.mean()
print("\nMean values of each column:")
print(mean_values)
median_values=df.median()
print("\nMedian values of each column:")
print(median_values)
mode_value=df.mode().iloc[0]
print("\nMode values of each column:")
print(mode_value)
```

```python
import pandas as pd
import seaborn as sns
from statistics import mode

tips = sns.load_dataset('tips')
df=pd.DataFrame(tips)
print("First 5 rows of dataset:")
print(df.head())
mean_values=df.mean(numeric_only=True)
print("\nMean:")
print(mean_values)
median_values=df.median(numeric_only=True)
print("\nMedian:")
print(median_values)
mode_value=df.mode(numeric_only=True).iloc[0]
print("\nMode:")
print(mode_value)
```

```python
import pandas as pd
from sklearn.datasets import load_iris
iris=load_iris()
df=pd.DataFrame(iris.data, columns=iris.feature_names)
print("First 5 rows of dataset:")
print(df.head())
range_values=df.max()-df.min()
print("\nMean values:")
print(range_values)
variance_values= df.var()
print("\nVariance:")
print(variance_values)
standard_deviation_values=df.std()
print("\nStandard deviation values:")
print(standard_deviation_values)
Q1=df.quantile(0.25)
Q3=df.quantile(0.75)
IQR=Q3-Q1
print("\nInterquartile range:")
print(IQR)


```

```python
import pandas as pd
import seaborn as sns
tips = sns.load_dataset('tips')
df=pd.DataFrame(tips)
print("First 5 rows of dataset:")
print(df.head())
range_values=df.max(numeric_only=True)-df.min(numeric_only=True)
print("\nRange values:")
print(range_values)
variance_values= df.var(numeric_only=True)
print("\nVariance:")
print(variance_values)
standard_deviation_values=df.std(numeric_only=True)
print("\nStandard deviation values:")
print(standard_deviation_values)
Q1=df.quantile(0.25, numeric_only=True)
Q3=df.quantile(0.75, numeric_only=True)
IQR=Q3-Q1
print("\nInterquartile range:")
print(IQR)
```

```python
import pandas as pd
from sklearn.datasets import load_iris
iris=load_iris()
df=pd.DataFrame(iris.data, columns=iris.feature_names)
print("First 5 rows of dataset:")
print(df.head())
skewness_values=df.skew()
print("\nSkewness values:")
print(skewness_values)
covariance_matrix=df.cov()
print("\nCovariance matrix:")
print(covariance_matrix)
```
